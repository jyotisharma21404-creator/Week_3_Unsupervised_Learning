```python
# Part 1- Perform K-Means on Iris Dataset and Visualize Clusters
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()
X = iris.data
print("Dataset shape:", X.shape)
```

```python
print("Feature names:")
print(iris.feature_names)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled[: 5])
```

```python
kmeans = KMeans(
    n_clusters = 3,
    random_state = 4,
    n_init = 10
)

clusters = kmeans.fit_predict(X_scaled)

print("Cluster labels:")
print(clusters)
```

```python
plt.figure(figsize = (8, 5))
plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c = clusters,
    s = 50
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("K-Means Clustering on Iris Dataset")

plt.show()
```

```python
# Part 2: Apply PCA to Reduce Dataset Dimensions
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
iris = load_iris()
X = iris.data
y = iris.target
print("Original shape:", X.shape)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled.shape)
```

```python
pca = PCA(n_components = 2)
X_pca = pca.fit_transform(X_scaled)
print("Original shape:", X.shape)
print("Reduced shape:", X_pca.shape)
```

```python
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)
print("Total Variance Exaplained:")
print(sum(pca.explained_variance_ratio_))
```

```python
plt.figure(figsize = (8,5))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c = y, s = 50)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA of Iris Dataset")

plt.show()
```

```python
# Mini Project 3- Iris Flower Clustering
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
iris = load_iris()
X = iris.data
y = iris.target
print("Dataset shape:", X.shape)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(
    n_clusters = 3,
    random_state = 42,
    n_init = 10
)
predicted_clusters = kmeans.fit_predict(X_scaled)
print(predicted_clusters)
```

```python
df = pd.DataFrame(
    X,
    columns = iris.feature_names
)
df["True Label"] = y
df["Predicted Cluster"] = predicted_clusters
df.head()
```

```python
plt.figure(figsize = (8, 5))
plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c = predicted_clusters,
    s = 50
)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Iris Flower Clustering using K-Means")

plt.show()
```

```python
comparision = pd.crosstab(
    df["True Label"],
    df["Predicted Cluster"]
)
print(comparision)
```

```python
print("Number of clusters:", kmeans.n_clusters)
print("Clustering completed successfully!")
```
